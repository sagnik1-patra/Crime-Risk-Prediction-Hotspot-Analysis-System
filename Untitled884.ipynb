```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

warnings.filterwarnings("ignore")

# ============================================================
# PATH SETTINGS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Crime Risk Prediction & Hotspot Analysis System"
CSV_PATH = os.path.join(BASE_DIR, "NCRB_CII-2020_Table.No-19B.2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

PREFIX = "pso_"

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

city_col = "City (Col. 2)"
target_col = "Total - Total Persons Arrested by age and Sex - (Col. 20)"

df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df = df.dropna()

# ============================================================
# CREATE TARGET
# ============================================================

df["Crime_Risk_Category"] = pd.qcut(
    df[target_col],
    q=3,
    labels=["Low Risk", "Medium Risk", "High Risk"],
    duplicates="drop"
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

df["Male_Female_Ratio"] = df["Total - Male - (Col. 18)"] / (
    df["Total - Female - (Col. 19)"] + 1
)

df["Juvenile_Percentage"] = (
    df["Juveniles Apprehended - Total - (Col. 5)"] / (df[target_col] + 1)
) * 100

df["Youth_18_30_Percentage"] = (
    df["18 Years & Above - Below 30 Years - Total - (Col. 8)"] / (df[target_col] + 1)
) * 100

df["Senior_60_Plus_Percentage"] = (
    df["60 Years & Above - Total - (Col. 17)"] / (df[target_col] + 1)
) * 100

drop_cols = [
    "Si. No. (Col. 1)",
    city_col,
    target_col,
    "Crime_Risk_Category"
]

feature_cols = [col for col in df.columns if col not in drop_cols]

X = df[feature_cols]
y = df["Crime_Risk_Category"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded
)

# ============================================================
# PSO FEATURE SELECTION
# ============================================================

def evaluate_particle(position):
    mask = position > 0.5
    selected = np.where(mask == 1)[0]

    if len(selected) == 0:
        return 0

    model = RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        max_depth=5
    )

    model.fit(X_train[:, selected], y_train)
    pred = model.predict(X_test[:, selected])

    acc = accuracy_score(y_test, pred)
    penalty = len(selected) / len(position)

    fitness = acc - (0.01 * penalty)
    return fitness


def particle_swarm_optimization(
    n_particles=20,
    iterations=20,
    w=0.7,
    c1=1.5,
    c2=1.5
):
    n_features = X_train.shape[1]

    positions = np.random.rand(n_particles, n_features)
    velocities = np.random.rand(n_particles, n_features)

    personal_best_positions = positions.copy()
    personal_best_scores = np.array([
        evaluate_particle(p)
        for p in positions
    ])

    global_best_index = np.argmax(personal_best_scores)
    global_best_position = personal_best_positions[global_best_index].copy()
    global_best_score = personal_best_scores[global_best_index]

    progress = []

    for iteration in range(iterations):
        for i in range(n_particles):
            r1 = np.random.rand(n_features)
            r2 = np.random.rand(n_features)

            velocities[i] = (
                w * velocities[i]
                + c1 * r1 * (personal_best_positions[i] - positions[i])
                + c2 * r2 * (global_best_position - positions[i])
            )

            positions[i] = positions[i] + velocities[i]
            positions[i] = 1 / (1 + np.exp(-positions[i]))

            score = evaluate_particle(positions[i])

            if score > personal_best_scores[i]:
                personal_best_scores[i] = score
                personal_best_positions[i] = positions[i].copy()

        best_index = np.argmax(personal_best_scores)

        if personal_best_scores[best_index] > global_best_score:
            global_best_score = personal_best_scores[best_index]
            global_best_position = personal_best_positions[best_index].copy()

        progress.append(global_best_score)

        print(f"PSO Iteration {iteration + 1}/{iterations} | Best Fitness: {global_best_score:.4f}")

    return global_best_position, global_best_score, progress


best_position, best_fitness, pso_progress = particle_swarm_optimization()

best_mask = best_position > 0.5
selected_indices = np.where(best_mask == 1)[0]

if len(selected_indices) == 0:
    selected_indices = np.array([np.argmax(best_position)])

selected_features = [feature_cols[i] for i in selected_indices]

X_train_pso = X_train[:, selected_indices]
X_test_pso = X_test[:, selected_indices]
X_scaled_pso = X_scaled[:, selected_indices]

# ============================================================
# PSO RANDOM FOREST MODEL
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train_pso, y_train)
rf_pred = rf_model.predict(X_test_pso)
rf_accuracy = accuracy_score(y_test, rf_pred)

# ============================================================
# PSO GRADIENT BOOSTING MODEL
# ============================================================

gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train_pso, y_train)
gb_pred = gb_model.predict(X_test_pso)
gb_accuracy = accuracy_score(y_test, gb_pred)

# ============================================================
# PSO DEEP LEARNING MODEL
# ============================================================

num_classes = len(np.unique(y_encoded))

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_pso.shape[1],)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

dl_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = dl_model.fit(
    X_train_pso,
    y_train,
    validation_data=(X_test_pso, y_test),
    epochs=100,
    batch_size=4,
    verbose=1
)

dl_loss, dl_accuracy = dl_model.evaluate(X_test_pso, y_test, verbose=0)

# ============================================================
# RESULT CSV
# ============================================================

result_df = pd.DataFrame({
    "Model": [
        "PSO Random Forest",
        "PSO Gradient Boosting",
        "PSO Deep Learning"
    ],
    "Accuracy": [
        rf_accuracy,
        gb_accuracy,
        dl_accuracy
    ]
})

result_df.to_csv(os.path.join(BASE_DIR, PREFIX + "result.csv"), index=False)

# ============================================================
# PREDICTION CSV
# ============================================================

df[PREFIX + "Predicted_Risk_Label"] = encoder.inverse_transform(
    rf_model.predict(X_scaled_pso)
)

df[PREFIX + "Risk_Score"] = df[target_col]

df.to_csv(os.path.join(BASE_DIR, PREFIX + "prediction.csv"), index=False)

# ============================================================
# HOTSPOT CSV
# ============================================================

hotspot_df = df[[city_col, target_col, PREFIX + "Predicted_Risk_Label"]].sort_values(
    by=target_col,
    ascending=False
)

hotspot_df.to_csv(os.path.join(BASE_DIR, PREFIX + "hotspot_result.csv"), index=False)

# ============================================================
# SAVE MODELS
# ============================================================

dl_model.save(os.path.join(BASE_DIR, PREFIX + "model.h5"))

with open(os.path.join(BASE_DIR, PREFIX + "rf_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, PREFIX + "gb_model.pkl"), "wb") as f:
    pickle.dump(gb_model, f)

with open(os.path.join(BASE_DIR, PREFIX + "scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, PREFIX + "encoder.pkl"), "wb") as f:
    pickle.dump(encoder, f)

with open(os.path.join(BASE_DIR, PREFIX + "selected_features.pkl"), "wb") as f:
    pickle.dump(selected_features, f)

# ============================================================
# YAML CONFIG
# ============================================================

config = {
    "project_name": "Crime Risk Prediction & Hotspot Analysis System",
    "algorithm": "Particle Swarm Optimization Feature Selection",
    "prefix": PREFIX,
    "dataset_path": CSV_PATH,
    "target_column": target_col,
    "city_column": city_col,
    "selected_features": selected_features,
    "total_features": len(feature_cols),
    "selected_feature_count": len(selected_features),
    "models": [
        "PSO Random Forest",
        "PSO Gradient Boosting",
        "PSO Deep Learning"
    ],
    "output_directory": BASE_DIR
}

with open(os.path.join(BASE_DIR, PREFIX + "config.yaml"), "w") as f:
    yaml.dump(config, f, sort_keys=False)

# ============================================================
# JSON SUMMARY
# ============================================================

summary = {
    "rows": int(df.shape[0]),
    "columns": int(df.shape[1]),
    "algorithm": "Particle Swarm Optimization",
    "best_pso_fitness": float(best_fitness),
    "selected_features": selected_features,
    "selected_feature_count": int(len(selected_features)),
    "pso_random_forest_accuracy": float(rf_accuracy),
    "pso_gradient_boosting_accuracy": float(gb_accuracy),
    "pso_deep_learning_accuracy": float(dl_accuracy),
    "classes": list(encoder.classes_),
    "classification_report": classification_report(
        y_test,
        rf_pred,
        target_names=encoder.classes_,
        output_dict=True
    )
}

with open(os.path.join(BASE_DIR, PREFIX + "summary.json"), "w") as f:
    json.dump(summary, f, indent=4)

# ============================================================
# GRAPHS
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(pso_progress) + 1), pso_progress, marker="o")
plt.title("PSO Optimization Progress")
plt.xlabel("Iteration")
plt.ylabel("Best Fitness")
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "optimization_progress_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("PSO Model Accuracy Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "accuracy_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(result_df["Model"], result_df["Accuracy"], marker="o")
plt.title("PSO Model Comparison Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True)
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "comparison_graph.png"))
plt.show()

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(7, 5))
plt.imshow(cm)
plt.title("PSO Crime Risk Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), encoder.classes_, rotation=45)
plt.yticks(range(num_classes), encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "heatmap.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("PSO Crime Risk Result Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy Score")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "result_graph.png"))
plt.show()

risk_count = df[PREFIX + "Predicted_Risk_Label"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(risk_count.index, risk_count.values)
plt.title("PSO Predicted Crime Risk Category Graph")
plt.xlabel("Risk Category")
plt.ylabel("Number of Cities")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "prediction_graph.png"))
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(hotspot_df[city_col], hotspot_df[target_col])
plt.title("PSO Crime Hotspot Ranking Graph")
plt.xlabel("City")
plt.ylabel("Total Persons Arrested")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "hotspot_result_graph.png"))
plt.show()

feature_importance = pd.DataFrame({
    "Feature": selected_features,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

feature_importance.to_csv(
    os.path.join(BASE_DIR, PREFIX + "feature_importance.csv"),
    index=False
)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance["Feature"], feature_importance["Importance"])
plt.title("PSO Selected Feature Importance Graph")
plt.xlabel("Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "feature_importance_graph.png"))
plt.show()

print("\nAll PSO results saved successfully in:")
print(BASE_DIR)

print("\nGenerated PSO Files:")
print(PREFIX + "model.h5")
print(PREFIX + "rf_model.pkl")
print(PREFIX + "gb_model.pkl")
print(PREFIX + "scaler.pkl")
print(PREFIX + "encoder.pkl")
print(PREFIX + "selected_features.pkl")
print(PREFIX + "config.yaml")
print(PREFIX + "summary.json")
print(PREFIX + "result.csv")
print(PREFIX + "prediction.csv")
print(PREFIX + "hotspot_result.csv")
print(PREFIX + "feature_importance.csv")
print(PREFIX + "optimization_progress_graph.png")
print(PREFIX + "accuracy_graph.png")
print(PREFIX + "comparison_graph.png")
print(PREFIX + "heatmap.png")
print(PREFIX + "result_graph.png")
print(PREFIX + "prediction_graph.png")
print(PREFIX + "hotspot_result_graph.png")
print(PREFIX + "feature_importance_graph.png")
```
