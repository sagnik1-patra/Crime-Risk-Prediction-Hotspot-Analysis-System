```python
import os
import json
import yaml
import pickle
import warnings
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

warnings.filterwarnings("ignore")

# ============================================================
# PATH SETTINGS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Crime Risk Prediction & Hotspot Analysis System"
CSV_PATH = os.path.join(BASE_DIR, "NCRB_CII-2020_Table.No-19B.2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

PREFIX = "csa_"

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

city_col = "City (Col. 2)"
target_col = "Total - Total Persons Arrested by age and Sex - (Col. 20)"

df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df = df.dropna()

# ============================================================
# TARGET CREATION
# ============================================================

df["Crime_Risk_Category"] = pd.qcut(
    df[target_col],
    q=3,
    labels=["Low Risk", "Medium Risk", "High Risk"],
    duplicates="drop"
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

df["Male_Female_Ratio"] = df["Total - Male - (Col. 18)"] / (
    df["Total - Female - (Col. 19)"] + 1
)

df["Juvenile_Percentage"] = (
    df["Juveniles Apprehended - Total - (Col. 5)"] / (df[target_col] + 1)
) * 100

df["Youth_18_30_Percentage"] = (
    df["18 Years & Above - Below 30 Years - Total - (Col. 8)"] / (df[target_col] + 1)
) * 100

df["Senior_60_Plus_Percentage"] = (
    df["60 Years & Above - Total - (Col. 17)"] / (df[target_col] + 1)
) * 100

drop_cols = [
    "Si. No. (Col. 1)",
    city_col,
    target_col,
    "Crime_Risk_Category"
]

feature_cols = [col for col in df.columns if col not in drop_cols]

X = df[feature_cols]
y = df["Crime_Risk_Category"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded
)

# ============================================================
# CSA FEATURE SELECTION
# ============================================================

def evaluate_solution(mask):
    selected = np.where(mask == 1)[0]

    if len(selected) == 0:
        return 0

    model = RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        max_depth=5
    )

    model.fit(X_train[:, selected], y_train)
    pred = model.predict(X_test[:, selected])

    acc = accuracy_score(y_test, pred)
    penalty = len(selected) / len(mask)

    fitness = acc - (0.01 * penalty)
    return fitness


def initialize_crows(pop_size, n_features):
    return np.random.randint(0, 2, size=(pop_size, n_features))


def crow_search_algorithm(
    pop_size=20,
    iterations=20,
    awareness_probability=0.1,
    flight_length=2
):
    n_features = X_train.shape[1]

    crows = initialize_crows(pop_size, n_features)
    memory = crows.copy()

    memory_fitness = np.array([
        evaluate_solution(m)
        for m in memory
    ])

    best_index = np.argmax(memory_fitness)
    best_solution = memory[best_index].copy()
    best_fitness = memory_fitness[best_index]

    progress = []

    for iteration in range(iterations):

        new_crows = []

        for i in range(pop_size):
            random_crow = random.randint(0, pop_size - 1)

            if random.random() > awareness_probability:
                new_position = crows[i].copy()

                for j in range(n_features):
                    if random.random() < 0.5:
                        if random.random() < flight_length / n_features:
                            new_position[j] = memory[random_crow][j]
                        else:
                            new_position[j] = crows[i][j]
            else:
                new_position = np.random.randint(0, 2, n_features)

            if np.sum(new_position) == 0:
                new_position[random.randint(0, n_features - 1)] = 1

            new_crows.append(new_position)

        crows = np.array(new_crows)

        current_fitness = np.array([
            evaluate_solution(c)
            for c in crows
        ])

        for i in range(pop_size):
            if current_fitness[i] > memory_fitness[i]:
                memory[i] = crows[i].copy()
                memory_fitness[i] = current_fitness[i]

        best_index = np.argmax(memory_fitness)

        if memory_fitness[best_index] > best_fitness:
            best_fitness = memory_fitness[best_index]
            best_solution = memory[best_index].copy()

        progress.append(best_fitness)

        print(f"CSA Iteration {iteration + 1}/{iterations} | Best Fitness: {best_fitness:.4f}")

    return best_solution, best_fitness, progress


best_mask, best_fitness, csa_progress = crow_search_algorithm()

selected_indices = np.where(best_mask == 1)[0]
selected_features = [feature_cols[i] for i in selected_indices]

X_train_csa = X_train[:, selected_indices]
X_test_csa = X_test[:, selected_indices]
X_scaled_csa = X_scaled[:, selected_indices]

# ============================================================
# CSA RANDOM FOREST MODEL
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train_csa, y_train)
rf_pred = rf_model.predict(X_test_csa)
rf_accuracy = accuracy_score(y_test, rf_pred)

# ============================================================
# CSA GRADIENT BOOSTING MODEL
# ============================================================

gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train_csa, y_train)
gb_pred = gb_model.predict(X_test_csa)
gb_accuracy = accuracy_score(y_test, gb_pred)

# ============================================================
# CSA DEEP LEARNING MODEL
# ============================================================

num_classes = len(np.unique(y_encoded))

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_csa.shape[1],)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

dl_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = dl_model.fit(
    X_train_csa,
    y_train,
    validation_data=(X_test_csa, y_test),
    epochs=100,
    batch_size=4,
    verbose=1
)

dl_loss, dl_accuracy = dl_model.evaluate(X_test_csa, y_test, verbose=0)

# ============================================================
# RESULT CSV
# ============================================================

result_df = pd.DataFrame({
    "Model": [
        "CSA Random Forest",
        "CSA Gradient Boosting",
        "CSA Deep Learning"
    ],
    "Accuracy": [
        rf_accuracy,
        gb_accuracy,
        dl_accuracy
    ]
})

result_df.to_csv(os.path.join(BASE_DIR, PREFIX + "result.csv"), index=False)

# ============================================================
# PREDICTION CSV
# ============================================================

df[PREFIX + "Predicted_Risk_Label"] = encoder.inverse_transform(
    rf_model.predict(X_scaled_csa)
)

df[PREFIX + "Risk_Score"] = df[target_col]

df.to_csv(os.path.join(BASE_DIR, PREFIX + "prediction.csv"), index=False)

# ============================================================
# HOTSPOT CSV
# ============================================================

hotspot_df = df[[city_col, target_col, PREFIX + "Predicted_Risk_Label"]].sort_values(
    by=target_col,
    ascending=False
)

hotspot_df.to_csv(os.path.join(BASE_DIR, PREFIX + "hotspot_result.csv"), index=False)

# ============================================================
# SAVE MODELS
# ============================================================

dl_model.save(os.path.join(BASE_DIR, PREFIX + "model.h5"))

with open(os.path.join(BASE_DIR, PREFIX + "rf_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, PREFIX + "gb_model.pkl"), "wb") as f:
    pickle.dump(gb_model, f)

with open(os.path.join(BASE_DIR, PREFIX + "scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, PREFIX + "encoder.pkl"), "wb") as f:
    pickle.dump(encoder, f)

with open(os.path.join(BASE_DIR, PREFIX + "selected_features.pkl"), "wb") as f:
    pickle.dump(selected_features, f)

# ============================================================
# YAML CONFIG
# ============================================================

config = {
    "project_name": "Crime Risk Prediction & Hotspot Analysis System",
    "algorithm": "Crow Search Algorithm Feature Selection",
    "prefix": PREFIX,
    "dataset_path": CSV_PATH,
    "target_column": target_col,
    "city_column": city_col,
    "selected_features": selected_features,
    "total_features": len(feature_cols),
    "selected_feature_count": len(selected_features),
    "models": [
        "CSA Random Forest",
        "CSA Gradient Boosting",
        "CSA Deep Learning"
    ],
    "output_directory": BASE_DIR
}

with open(os.path.join(BASE_DIR, PREFIX + "config.yaml"), "w") as f:
    yaml.dump(config, f, sort_keys=False)

# ============================================================
# JSON SUMMARY
# ============================================================

summary = {
    "rows": int(df.shape[0]),
    "columns": int(df.shape[1]),
    "algorithm": "Crow Search Algorithm",
    "best_csa_fitness": float(best_fitness),
    "selected_features": selected_features,
    "selected_feature_count": int(len(selected_features)),
    "csa_random_forest_accuracy": float(rf_accuracy),
    "csa_gradient_boosting_accuracy": float(gb_accuracy),
    "csa_deep_learning_accuracy": float(dl_accuracy),
    "classes": list(encoder.classes_),
    "classification_report": classification_report(
        y_test,
        rf_pred,
        target_names=encoder.classes_,
        output_dict=True
    )
}

with open(os.path.join(BASE_DIR, PREFIX + "summary.json"), "w") as f:
    json.dump(summary, f, indent=4)

# ============================================================
# GRAPHS
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(csa_progress) + 1), csa_progress, marker="o")
plt.title("CSA Optimization Progress")
plt.xlabel("Iteration")
plt.ylabel("Best Fitness")
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "optimization_progress_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("CSA Model Accuracy Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "accuracy_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(result_df["Model"], result_df["Accuracy"], marker="o")
plt.title("CSA Model Comparison Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True)
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "comparison_graph.png"))
plt.show()

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(7, 5))
plt.imshow(cm)
plt.title("CSA Crime Risk Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), encoder.classes_, rotation=45)
plt.yticks(range(num_classes), encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "heatmap.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("CSA Crime Risk Result Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy Score")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "result_graph.png"))
plt.show()

risk_count = df[PREFIX + "Predicted_Risk_Label"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(risk_count.index, risk_count.values)
plt.title("CSA Predicted Crime Risk Category Graph")
plt.xlabel("Risk Category")
plt.ylabel("Number of Cities")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "prediction_graph.png"))
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(hotspot_df[city_col], hotspot_df[target_col])
plt.title("CSA Crime Hotspot Ranking Graph")
plt.xlabel("City")
plt.ylabel("Total Persons Arrested")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "hotspot_result_graph.png"))
plt.show()

feature_importance = pd.DataFrame({
    "Feature": selected_features,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

feature_importance.to_csv(
    os.path.join(BASE_DIR, PREFIX + "feature_importance.csv"),
    index=False
)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance["Feature"], feature_importance["Importance"])
plt.title("CSA Selected Feature Importance Graph")
plt.xlabel("Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "feature_importance_graph.png"))
plt.show()

print("\nAll CSA results saved successfully in:")
print(BASE_DIR)
```
