```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

warnings.filterwarnings("ignore")

# =========================
# PATH
# =========================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Crime Risk Prediction & Hotspot Analysis System"
CSV_PATH = os.path.join(BASE_DIR, "NCRB_CII-2020_Table.No-19B.2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

# =========================
# LOAD DATA
# =========================

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

city_col = "City (Col. 2)"
target_col = "Total - Total Persons Arrested by age and Sex - (Col. 20)"

df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df = df.dropna()

# =========================
# CREATE RISK CATEGORY
# =========================

df["Crime_Risk_Category"] = pd.qcut(
    df[target_col],
    q=3,
    labels=["Low Risk", "Medium Risk", "High Risk"],
    duplicates="drop"
)

# =========================
# FEATURE ENGINEERING
# =========================

df["Male_Female_Ratio"] = df["Total - Male - (Col. 18)"] / (df["Total - Female - (Col. 19)"] + 1)

df["Juvenile_Percentage"] = (
    df["Juveniles Apprehended - Total - (Col. 5)"] / (df[target_col] + 1)
) * 100

df["Youth_18_30_Percentage"] = (
    df["18 Years & Above - Below 30 Years - Total - (Col. 8)"] / (df[target_col] + 1)
) * 100

df["Senior_60_Plus_Percentage"] = (
    df["60 Years & Above - Total - (Col. 17)"] / (df[target_col] + 1)
) * 100

# =========================
# FEATURES AND TARGET
# =========================

drop_cols = [
    "Si. No. (Col. 1)",
    city_col,
    target_col,
    "Crime_Risk_Category"
]

feature_cols = [col for col in df.columns if col not in drop_cols]

X = df[feature_cols]
y = df["Crime_Risk_Category"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded
)

# =========================
# MODEL 1: RANDOM FOREST
# =========================

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

# =========================
# MODEL 2: GRADIENT BOOSTING
# =========================

gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train, y_train)
gb_pred = gb_model.predict(X_test)
gb_accuracy = accuracy_score(y_test, gb_pred)

# =========================
# MODEL 3: DEEP LEARNING
# =========================

num_classes = len(np.unique(y_encoded))

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

dl_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = dl_model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=100,
    batch_size=4,
    verbose=1
)

dl_loss, dl_accuracy = dl_model.evaluate(X_test, y_test, verbose=0)
dl_pred = np.argmax(dl_model.predict(X_test), axis=1)

# =========================
# PREDICTION CSV
# =========================

df["Predicted_Risk_Label"] = encoder.inverse_transform(rf_model.predict(X_scaled))
df["Risk_Score"] = df[target_col]

prediction_csv = os.path.join(BASE_DIR, "crime_risk_prediction.csv")
df.to_csv(prediction_csv, index=False)

# =========================
# RESULT CSV
# =========================

result_df = pd.DataFrame({
    "Model": ["Random Forest", "Gradient Boosting", "Deep Learning"],
    "Accuracy": [rf_accuracy, gb_accuracy, dl_accuracy]
})

result_csv = os.path.join(BASE_DIR, "crime_risk_result.csv")
result_df.to_csv(result_csv, index=False)

# =========================
# SAVE H5, PKL, YAML, JSON
# =========================

dl_model.save(os.path.join(BASE_DIR, "crime_risk_model.h5"))

with open(os.path.join(BASE_DIR, "crime_risk_rf_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, "crime_risk_gb_model.pkl"), "wb") as f:
    pickle.dump(gb_model, f)

with open(os.path.join(BASE_DIR, "crime_risk_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, "crime_risk_encoder.pkl"), "wb") as f:
    pickle.dump(encoder, f)

config = {
    "project_name": "Crime Risk Prediction & Hotspot Analysis System",
    "dataset_path": CSV_PATH,
    "target_column": target_col,
    "city_column": city_col,
    "features_used": feature_cols,
    "models": ["Random Forest", "Gradient Boosting", "Deep Learning"],
    "output_directory": BASE_DIR
}

with open(os.path.join(BASE_DIR, "crime_risk_config.yaml"), "w") as f:
    yaml.dump(config, f, sort_keys=False)

summary = {
    "rows": int(df.shape[0]),
    "columns": int(df.shape[1]),
    "random_forest_accuracy": float(rf_accuracy),
    "gradient_boosting_accuracy": float(gb_accuracy),
    "deep_learning_accuracy": float(dl_accuracy),
    "classes": list(encoder.classes_),
    "classification_report": classification_report(
        y_test,
        rf_pred,
        target_names=encoder.classes_,
        output_dict=True
    )
}

with open(os.path.join(BASE_DIR, "crime_risk_summary.json"), "w") as f:
    json.dump(summary, f, indent=4)

# =========================
# ACCURACY GRAPH
# =========================

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("Model Accuracy Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_accuracy_graph.png"))
plt.show()

# =========================
# COMPARISON GRAPH
# =========================

plt.figure(figsize=(8, 5))
plt.plot(result_df["Model"], result_df["Accuracy"], marker="o")
plt.title("Model Comparison Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_comparison_graph.png"))
plt.show()

# =========================
# HEATMAP / CONFUSION MATRIX
# =========================

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(7, 5))
plt.imshow(cm)
plt.title("Crime Risk Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), encoder.classes_, rotation=45)
plt.yticks(range(num_classes), encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_heatmap.png"))
plt.show()

# =========================
# RESULT GRAPH
# =========================

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("Crime Risk Result Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy Score")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_result_graph.png"))
plt.show()

# =========================
# PREDICTION GRAPH
# =========================

risk_count = df["Predicted_Risk_Label"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(risk_count.index, risk_count.values)
plt.title("Predicted Crime Risk Category Graph")
plt.xlabel("Risk Category")
plt.ylabel("Number of Cities")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_prediction_graph.png"))
plt.show()

# =========================
# HOTSPOT GRAPH
# =========================

hotspot_df = df[[city_col, target_col, "Predicted_Risk_Label"]].sort_values(
    by=target_col,
    ascending=False
)

hotspot_csv = os.path.join(BASE_DIR, "crime_hotspot_result.csv")
hotspot_df.to_csv(hotspot_csv, index=False)

plt.figure(figsize=(12, 6))
plt.bar(hotspot_df[city_col], hotspot_df[target_col])
plt.title("Crime Hotspot Ranking Graph")
plt.xlabel("City")
plt.ylabel("Total Persons Arrested")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_hotspot_result_graph.png"))
plt.show()

print("\nAll results saved successfully in:")
print(BASE_DIR)

print("\nGenerated Files:")
print("crime_risk_model.h5")
print("crime_risk_rf_model.pkl")
print("crime_risk_gb_model.pkl")
print("crime_risk_scaler.pkl")
print("crime_risk_encoder.pkl")
print("crime_risk_config.yaml")
print("crime_risk_summary.json")
print("crime_risk_result.csv")
print("crime_risk_prediction.csv")
print("crime_hotspot_result.csv")
print("crime_risk_accuracy_graph.png")
print("crime_risk_heatmap.png")
print("crime_risk_comparison_graph.png")
print("crime_risk_result_graph.png")
print("crime_risk_prediction_graph.png")
print("crime_hotspot_result_graph.png")
```
