```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# PATH SETTINGS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Crime Risk Prediction & Hotspot Analysis System"

CSV_PATH = os.path.join(BASE_DIR, "NCRB_CII-2020_Table.No-19B.2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv(CSV_PATH)

print("Dataset Loaded Successfully")
print("Shape:", df.shape)
print(df.head())

# ============================================================
# CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.replace("\n", " ")
    .str.replace("  ", " ")
)

city_col = "City (Col. 2)"
target_col = "Total - Total Persons Arrested by age and Sex - (Col. 20)"

# Remove serial number column
if "Si. No. (Col. 1)" in df.columns:
    df = df.drop(columns=["Si. No. (Col. 1)"])

# ============================================================
# CREATE RISK CATEGORY TARGET
# ============================================================

df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df = df.dropna()

df["Crime_Risk_Category"] = pd.qcut(
    df[target_col],
    q=3,
    labels=["Low Risk", "Medium Risk", "High Risk"],
    duplicates="drop"
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

if target_col in numeric_cols:
    numeric_cols.remove(target_col)

df["Male_Female_Ratio"] = (
    df["Total - Male - (Col. 18)"] /
    (df["Total - Female - (Col. 19)"] + 1)
)

df["Juvenile_Percentage"] = (
    df["Juveniles Apprehended - Total - (Col. 5)"] /
    (df[target_col] + 1)
) * 100

df["Youth_18_30_Percentage"] = (
    df["18 Years & Above - Below 30 Years - Total - (Col. 8)"] /
    (df[target_col] + 1)
) * 100

df["Senior_60_Plus_Percentage"] = (
    df["60 Years & Above - Total - (Col. 17)"] /
    (df[target_col] + 1)
) * 100

# Refresh numeric columns
feature_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
feature_cols.remove(target_col)

X = df[feature_cols]
y = df["Crime_Risk_Category"]

# ============================================================
# ENCODE TARGET
# ============================================================

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# ============================================================
# SCALE FEATURES
# ============================================================

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ============================================================
# TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded
)

# ============================================================
# RANDOM FOREST MODEL
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

print("\nRandom Forest Accuracy:", rf_accuracy)

# ============================================================
# DEEP LEARNING MODEL
# ============================================================

num_classes = len(np.unique(y_encoded))

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

dl_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = dl_model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=100,
    batch_size=4,
    callbacks=[early_stop],
    verbose=1
)

dl_loss, dl_accuracy = dl_model.evaluate(X_test, y_test, verbose=0)

print("\nDeep Learning Accuracy:", dl_accuracy)

# ============================================================
# FULL DATA PREDICTION
# ============================================================

df["Predicted_Risk_Label"] = label_encoder.inverse_transform(
    rf_model.predict(X_scaled)
)

df["Risk_Score"] = df[target_col]

prediction_path = os.path.join(BASE_DIR, "crime_risk_predictions.csv")
df.to_csv(prediction_path, index=False)

# ============================================================
# SAVE FILES
# ============================================================

# H5 model
h5_path = os.path.join(BASE_DIR, "crime_risk_deep_learning_model.h5")
dl_model.save(h5_path)

# PKL files
pkl_model_path = os.path.join(BASE_DIR, "crime_risk_random_forest_model.pkl")
pkl_scaler_path = os.path.join(BASE_DIR, "crime_risk_scaler.pkl")
pkl_encoder_path = os.path.join(BASE_DIR, "crime_risk_label_encoder.pkl")

with open(pkl_model_path, "wb") as f:
    pickle.dump(rf_model, f)

with open(pkl_scaler_path, "wb") as f:
    pickle.dump(scaler, f)

with open(pkl_encoder_path, "wb") as f:
    pickle.dump(label_encoder, f)

# YAML config
config = {
    "project_name": "Crime Risk Prediction & Hotspot Analysis System",
    "dataset": CSV_PATH,
    "target_column": target_col,
    "risk_target": "Crime_Risk_Category",
    "features_used": feature_cols,
    "models": {
        "machine_learning": "RandomForestClassifier",
        "deep_learning": "Dense Neural Network"
    },
    "outputs": {
        "h5_model": h5_path,
        "pkl_model": pkl_model_path,
        "pkl_scaler": pkl_scaler_path,
        "pkl_encoder": pkl_encoder_path,
        "prediction_csv": prediction_path
    }
}

yaml_path = os.path.join(BASE_DIR, "crime_risk_config.yaml")

with open(yaml_path, "w") as f:
    yaml.dump(config, f, sort_keys=False)

# JSON result summary
summary = {
    "dataset_rows": int(df.shape[0]),
    "dataset_columns": int(df.shape[1]),
    "random_forest_accuracy": float(rf_accuracy),
    "deep_learning_accuracy": float(dl_accuracy),
    "risk_classes": list(label_encoder.classes_),
    "classification_report": classification_report(
        y_test,
        rf_pred,
        target_names=label_encoder.classes_,
        output_dict=True
    )
}

json_path = os.path.join(BASE_DIR, "crime_risk_summary.json")

with open(json_path, "w") as f:
    json.dump(summary, f, indent=4)

# ============================================================
# GRAPHS
# ============================================================

# Accuracy graph
plt.figure(figsize=(8, 5))
plt.bar(["Random Forest", "Deep Learning"], [rf_accuracy, dl_accuracy])
plt.title("Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.savefig(os.path.join(BASE_DIR, "crime_risk_accuracy_graph.png"))
plt.close()

# Confusion matrix
cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(6, 5))
plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), label_encoder.classes_, rotation=45)
plt.yticks(range(num_classes), label_encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_confusion_matrix.png"))
plt.close()

# Feature importance graph
feature_importance = pd.DataFrame({
    "Feature": feature_cols,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

feature_importance_path = os.path.join(BASE_DIR, "crime_risk_feature_importance.csv")
feature_importance.to_csv(feature_importance_path, index=False)

plt.figure(figsize=(10, 6))
plt.barh(feature_importance["Feature"], feature_importance["Importance"])
plt.title("Feature Importance")
plt.xlabel("Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_risk_feature_importance.png"))
plt.close()

# Hotspot ranking graph
hotspot_df = df[[city_col, target_col, "Predicted_Risk_Label"]].sort_values(
    by=target_col,
    ascending=False
)

hotspot_path = os.path.join(BASE_DIR, "crime_hotspot_ranking.csv")
hotspot_df.to_csv(hotspot_path, index=False)

plt.figure(figsize=(12, 6))
plt.bar(hotspot_df[city_col], hotspot_df[target_col])
plt.title("Crime Hotspot Ranking by Total Persons Arrested")
plt.xlabel("City")
plt.ylabel("Total Persons Arrested")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crime_hotspot_ranking_graph.png"))
plt.close()

# ============================================================
# FINAL OUTPUT
# ============================================================

print("\nAll files saved successfully in:")
print(BASE_DIR)

print("\nGenerated Files:")
print("1. crime_risk_deep_learning_model.h5")
print("2. crime_risk_random_forest_model.pkl")
print("3. crime_risk_scaler.pkl")
print("4. crime_risk_label_encoder.pkl")
print("5. crime_risk_config.yaml")
print("6. crime_risk_summary.json")
print("7. crime_risk_predictions.csv")
print("8. crime_hotspot_ranking.csv")
print("9. crime_risk_accuracy_graph.png")
print("10. crime_risk_confusion_matrix.png")
print("11. crime_risk_feature_importance.png")
print("12. crime_hotspot_ranking_graph.png")
```
